```python
import pandas as pd


df = pd.read_csv("AdaBoostClassifier_random_state0.csv")

# Contar la cantidad de datasets únicos
dataset_ids = df['Dataset'].unique()
print("Cantidad de datasets únicos:", len(dataset_ids))

df= df.drop(columns=['source_csv','Classifier'])

print(df.columns)
print(len(df.columns))
```

```python
df_encoded = pd.get_dummies(
    df,
    columns=[
        "Imputer Strategy",
        "Categorical Strategy",
        "Feature Selection",
        "Scaler"
    ],
    dtype= int
)
print(df_encoded)
```

```python
groups = {
    "Imputer Strategy": "Imputer Strategy_",
    "Categorical Strategy": "Categorical Strategy_",
    "Feature Selection": "Feature Selection_",
    "Scaler": "Scaler_"
}

for name, prefix in groups.items():
    cols = [c for c in df_encoded.columns if c.startswith(prefix)]
    print(f"{name}: {len(cols)} variables")
```
